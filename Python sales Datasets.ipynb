```python
import pandas as pd # create a table in pyhton using pandas

data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010],

    "Product": ["Laptop", "Mouse", "Keyboard", "Laptop",
                "Monitor", "Mouse", "Keyboard", "Laptop",
                "Monitor", "Mouse"],

    "Category": ["Electronics", "Accessories", "Accessories",
                 "Electronics", "Electronics", "Accessories",
                 "Accessories", "Electronics", "Electronics",
                 "Accessories"],

    "Quantity": [2, 5, 3, 1, 2, 4, 2, 3, 1, 6],

    "Price": [50000, 500, 1500, 50000, 12000,
              500, 1500, 50000, 12000, 500]
}

df = pd.DataFrame(data)

df
```

```python
df["Revenue"]=df["Quantity"]*df["Price"] # R=Q*P

df
```

```python
total_revenue=df["Revenue"].sum() # to calculate the total column in the table 
print("Total Revenue: ₹",total_revenue)
```

```python
total_price=df["Price"].sum()
print("Total Price: ₹",total_price)
```

```python
product_sales=df.groupby("Product")["Revenue"].sum()
print(product_sales)
df
```

```python
import matplotlib.pyplot as plt

product_sales.plot(
    kind="bar",
    color=["skyblue","orange","green","red"],
    figsize=(10, 5)
)

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_quantity=df.groupby("Category")["Quantity"].sum()
print(category_quantity)
```

```python
import matplotlib.pyplot as plt

category_quantity.plot(
    kind="bar",
    color=["orange", "skyblue"],
    figsize=(10, 5)
)

plt.title("Quantity Sold by Category")
plt.xlabel("Category")
plt.ylabel("Units Sold")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
ax = category_quantity.plot(
    kind="bar",
    color=["orange", "skyblue"],
    figsize=(8, 5)
)

plt.title("Quantity Sold by Category")
plt.xlabel("Category")
plt.ylabel("Units Sold")
plt.xticks(rotation=0)

for bar in ax.patches:
    height = bar.get_height()
    ax.annotate(
        f"{height:.0f}",
        (bar.get_x() + bar.get_width() / 2, height),
        ha="center",
        va="bottom"
    )

plt.ylim(0, category_quantity.max() * 1.15)
plt.tight_layout()
plt.show()
```
